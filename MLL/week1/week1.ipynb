```python
#3. Find the positions of: elements in x where its value is 
#   more than its corresponding element in y, and elements in 
#   x where its value is equals to its corresponding element in y.

x = np.array([21, 64, 86, 22, 74, 55, 81, 79, 90, 89])
y = np.array([21, 7, 3, 45, 10, 29, 55, 4, 37, 18])

greater=np.where(x>y)
equal=np.where(x==y)

print(greater,equal)
```

```python
#2. Consider the following matrices :
#   A= ((1, 2, 3), (4, 5, 6), (7, 8, 10)) and B = ((7, 8, 10) ,(4, 5, 6), (1, 2, 3))
#   Write a python program to perform the following using Numeric Python (numpy).
A = np.array([[1, 2, 3],[4, 5, 6],[7, 8, 10]])

B = np.array([[7, 8, 10],[4, 5, 6],[1, 2, 3]])

# i) Addition and Subtraction
C = A + B
E = A - B

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nAddition (C = A + B):")
print(C)

print("\nSubtraction (E = A - B):")
print(E)

# ii) Sum of elements
print("\nSum of all elements of Matrix A:")
print(np.sum(A))

print("\nSum of each column of Matrix B:")
print(np.sum(B, axis=0))

print("\nSum of each row of Matrix C:")
print(np.sum(C, axis=1))

# iii) Matrix multiplication
D = np.matmul(A, B)

print("\nProduct of A and B (D):")
print(D)

# iv) Sort the elements of Matrix C
E = np.sort(C, axis=None).reshape(C.shape)

print("\nSorted Matrix C (E):")
print(E)

# v) Transpose of Matrix E
T = E.T

print("\nTranspose of Matrix E:")
print(T)
```
