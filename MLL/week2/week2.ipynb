```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

### 1. Follow along with these steps:
a) Create a figure object called fig using plt.figure()<br>
b) Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis ax.<br>
c) Plot (x,y) on that axes and set the labels and titles to match the plot below:<br>

```python
x = np.arange(0, 101)
y = 2 * x

fig = plt.figure()

ax = fig.add_axes([0, 0, 1, 1])

ax.plot(x, y)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('title')

plt.show()
```

### 2. Create a figure object and put two axes on it, ax1 and ax2. <br>Located at [0,0,1,1] and [0.2,0.5,.2,.2] respectively. Now plot (x,y) on both axes <br>and call your figure object to show it.

```python
fig = plt.figure()

x=np.arange(0,101)
y=2*x

ax1= fig.add_axes([0,0,1,1])
ax2=fig.add_axes([0.2,0.5,0.2,0.2])

ax1.plot(x,y)
ax2.plot(x,y)

plt.show()
```

### 3. Use the company sales dataset csv file, read Total profit of all months<br>and show it using a line plotTotal profit data provided for each month. <br>Generated line plot must include the following properties: –
#### a. X label name = Month Number<br>b. Y label name = Total profit

```python

df = pd.read_csv("company-sales.csv")

plt.plot(df["month_number"], df["total_profit"])

plt.xlabel("Month Number")
plt.ylabel("Total profit")
plt.title("Company Profit Per Month")

plt.show()
```

### 4. Use the company sales dataset csv file, get total profit of all months <br>and show line plot with the following Style properties. <br>Generated line plot must include following Style properties: –
#### a. Line Style dotted and Line-color should be red<br>b. Show legend at the lower right location.<br>c. X label name = Month Number<br>d. Y label name = Sold units number<br>e. Add a circle marker.<br>f. Line marker color as read<br>g. Line width should be 3

```python
plt.plot(df["month_number"],
         df["total_profit"],          # Replace with "Total Profit" if needed
         color='red',
         linestyle=':',
         marker='o',
         markerfacecolor='red',
         label='Profit Data',
         linewidth=3
         )

# Labels
plt.xlabel("Month Number")
plt.ylabel("Sold units number")

# Legend
plt.legend(loc='lower right')

plt.show()
```
